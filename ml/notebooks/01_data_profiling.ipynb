```python
import pandas as pd
import pymysql
import json
import time
from pymysql.err import OperationalError

# ==========================
# CONFIG
# ==========================
CSV_FILE = "/Users/kousiknandury/Desktop/Spotify_demo/ml/data/songs.csv"

DB_CONFIG = {
    "host": "127.0.0.1",
    "user": "root",
    "password": "Pokemon!12340987",
    "database": "music_platform",
    "charset": "utf8mb4",
    "autocommit": False,
}

BATCH_SIZE = 1000
MAX_RETRIES = 5
RETRY_BACKOFF_SEC = 2

# ==========================
# LOAD CSV
# ==========================
df = pd.read_csv(CSV_FILE)

# ==========================
# BUILD audio_features JSON
# ==========================
AUDIO_COLS = [
    "danceability", "energy", "loudness", "speechiness",
    "acousticness", "instrumentalness", "liveness",
    "valence", "tempo"
]

def build_audio_features(row):
    return json.dumps({
        k: float(row[k]) if pd.notna(row[k]) else None
        for k in AUDIO_COLS
    })

df["audio_features"] = df.apply(build_audio_features, axis=1)

# ==========================
# CLEAN & MAP TO SCHEMA
# ==========================
release_dates = pd.to_datetime(
    df["track_album_release_date"],
    errors="coerce"
).dt.date

final_df = pd.DataFrame({
    "track_id": df["track_id"],
    "track_name": df["track_name"],
    "artist": df["track_artist"],
    "album_id": df["track_album_id"],
    "album_name": df["track_album_name"],
    "release_date": release_dates.where(release_dates.notna(), None),
    "genre": df["playlist_genre"],
    "subgenre": df["playlist_subgenre"],
    "popularity": pd.to_numeric(df["track_popularity"], errors="coerce"),
    "duration_ms": pd.to_numeric(df["duration_ms"], errors="coerce"),
    "audio_features": df["audio_features"],
})

final_df = final_df.dropna(subset=["track_id"])
final_df = final_df.drop_duplicates(subset=["track_id"])

data = list(final_df.itertuples(index=False, name=None))

# ==========================
# SQL
# ==========================
SQL = """
INSERT INTO songs (
    track_id, track_name, artist, album_id, album_name,
    release_date, genre, subgenre, popularity, duration_ms, audio_features
)
VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)
ON DUPLICATE KEY UPDATE
    track_name = VALUES(track_name),
    artist = VALUES(artist),
    album_id = VALUES(album_id),
    album_name = VALUES(album_name),
    release_date = VALUES(release_date),
    genre = VALUES(genre),
    subgenre = VALUES(subgenre),
    popularity = VALUES(popularity),
    duration_ms = VALUES(duration_ms),
    audio_features = VALUES(audio_features)
"""

# ==========================
# RETRY-SAFE EXECUTION
# ==========================
def execute_batch(cursor, batch, attempt=1):
    try:
        cursor.executemany(SQL, batch)
        return True
    except OperationalError as e:
        if e.args[0] in (1205, 1213):  # lock wait / deadlock
            if attempt > MAX_RETRIES:
                raise
            wait = RETRY_BACKOFF_SEC * attempt
            print(f"⚠️  Lock issue (attempt {attempt}), retrying in {wait}s...")
            time.sleep(wait)
            return execute_batch(cursor, batch, attempt + 1)
        raise

# ==========================
# RUN IMPORT
# ==========================
conn = pymysql.connect(**DB_CONFIG)
cursor = conn.cursor()

cursor.execute("SET FOREIGN_KEY_CHECKS = 0")

inserted = 0

for i in range(0, len(data), BATCH_SIZE):
    batch = data[i:i+BATCH_SIZE]
    execute_batch(cursor, batch)
    conn.commit()
    inserted += len(batch)
    print(f"✅ Inserted {inserted}/{len(data)} rows")

cursor.execute("SET FOREIGN_KEY_CHECKS = 1")

cursor.close()
conn.close()

print("🎉 Import completed successfully")
```

```python
import pandas as pd
import json

# -------------------------
# LOAD ORIGINAL CSV
# -------------------------
df = pd.read_csv("/Users/kousiknandury/Desktop/Spotify_demo/ml/data/songs.csv")

# -------------------------
# BUILD audio_features JSON
# -------------------------
AUDIO_COLS = [
    "danceability", "energy", "loudness", "speechiness",
    "acousticness", "instrumentalness", "liveness",
    "valence", "tempo"
]

def build_audio_features(row):
    return json.dumps({
        k: float(row[k]) if pd.notna(row[k]) else None
        for k in AUDIO_COLS
    }, ensure_ascii=False)

df["audio_features"] = df.apply(build_audio_features, axis=1)

# -------------------------
# PARSE & CLEAN release_date
# -------------------------
release_dates = pd.to_datetime(
    df["track_album_release_date"],
    errors="coerce"
).dt.strftime("%Y-%m-%d")

# NaT → empty string (LOAD DATA will convert to NULL)
release_dates = release_dates.fillna("")

# -------------------------
# FINAL DATAFRAME (MATCHES TABLE)
# -------------------------
final_df = pd.DataFrame({
    "track_id": df["track_id"],
    "track_name": df["track_name"],
    "artist": df["track_artist"],
    "album_id": df["track_album_id"],
    "album_name": df["track_album_name"],
    "release_date": release_dates,
    "genre": df["playlist_genre"],
    "subgenre": df["playlist_subgenre"],
    "popularity": df["track_popularity"],
    "duration_ms": df["duration_ms"],
    "audio_features": df["audio_features"],
})

# Enforce PK constraints
final_df = final_df.dropna(subset=["track_id"])
final_df = final_df.drop_duplicates(subset=["track_id"])

# -------------------------
# WRITE MySQL-READY CSV
# -------------------------
final_df.to_csv(
    "songs_load.csv",
    index=False,
    encoding="utf-8",
    quoting=1,          # QUOTE_ALL
    lineterminator="\n"
)


print(f"✅ Generated songs_load.csv with {len(final_df)} rows")
```
